# Git + GitHub Cheat Sheet — Simple Flow (Mac + VS Code)

## 1. First-time setup for a new project

| Step | What to do | Command / notes |
|---|---|---|
| 1 | Open the project folder in VS Code | **File → Open Folder** |
| 2 | Open the terminal | **Terminal → New Terminal** |
| 3 | Check your location and files | `pwd` then `ls` |
| 4 | Check that Git is installed | `git --version` |
| 5 | Start Git tracking (once per project) | `git init` |
| 6 | Create `.gitignore` | Add `asi-venv/`, `.DS_Store`, `__pycache__/`, `*.pyc`, `.vscode/`, `.ipynb_checkpoints/` |
| 7 | Create an empty repository on GitHub | For the simplest setup, don't add a README, `.gitignore`, or license on GitHub |
| 8 | Connect the local project to GitHub | `git remote add origin https://github.com/USERNAME/REPO.git` |
| 9 | Name the branch `main` | `git branch -M main` |
| 10 | Check what Git sees | `git status` |
| 11 | Stage the files you want to save | `git add .` |
| 12 | Review staged files | `git status` |
| 13 | Create the first local commit | `git commit -m "Initial commit"` |
| 14 | Push to GitHub (first time) | `git push -u origin main` |

**Important:** Replace the example GitHub URL with your own repository URL. Do not upload virtual environments or secrets such as passwords/API keys.

## 2. Everyday workflow after setup

Use this flow whenever you edit or create files:

| Order | Action | Command |
|---:|---|---|
| 1 | Edit and save your files | In VS Code |
| 2 | See what changed | `git status` |
| 3 | Stage the changes you want to include | `git add .` |
| 4 | Review what is staged | `git status` |
| 5 | Save a checkpoint in Git's local history | `git commit -m "Describe the change"` |
| 6 | Upload committed changes to GitHub | `git push` |

### Remember the difference

| Command | Meaning |
|---|---|
| `git add .` | Stages changes for the next commit |
| `git commit -m "message"` | Saves a checkpoint **locally** |
| `git push` | Uploads commits to **GitHub** |
| `git status` | Shows changed/staged files |
| `git log --oneline` | Shows a short commit history |
| `git remote -v` | Shows the connected remote repository |
| `git branch` | Shows local branches |
| `git reset` | Unstages staged changes by default; it does not delete the files |

## 3. Quick troubleshooting

| Message / situation | What to do |
|---|---|
| `remote origin already exists` | Run `git remote -v`. If the URL is correct, don't add it again. |
| `nothing to commit, working tree clean` | Git sees no new changes to commit. Check that you saved your edits. |
| Virtual environment appears in `git status` | Make sure `.gitignore` contains `asi-venv/`. If the environment was already tracked, ignoring it alone won't untrack it. |
| `.DS_Store` appears | Add `.DS_Store` to `.gitignore`. |
| A subfolder contains its own `.git` | It may be a separate repository. Don't delete it unless you have confirmed you want to remove that nested repository. |
| Need to unstage files | `git reset` |

## 4. Useful checks

| Question | Command |
|---|---|
| Where am I? | `pwd` |
| What files are here? | `ls` |
| Is Git installed? | `git --version` |
| What changed? | `git status` |
| Which GitHub repository is connected? | `git remote -v` |
| What commits exist? | `git log --oneline` |
| Which branch am I on? | `git branch` |

## 5. One-time commands vs regular commands

| Usually once per project | Use during normal work |
|---|---|
| `git init` | `git status` |
| `git remote add origin ...` | `git add .` |
| `git branch -M main` | `git commit -m "message"` |
| First `git push -u origin main` | `git push` |

**Rule of thumb:** One project folder should normally have one Git repository. Don't run `git init` again in every subfolder.
