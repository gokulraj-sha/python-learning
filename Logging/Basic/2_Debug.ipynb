## **Debug**

```python
import  logging
logging.basicConfig(filename= 'test(debug).log', level=logging.DEBUG, force= True,
                    format = '%(asctime)s:%(levelname)s:%(message)s')


def add(a,b):
    return a+b

def sub(a,b):
    return a-b

def multi(a,b):
    return a*b

def devision(a,b):
    return a/b

num1 =10
num2 = 5


add_result = add(num1, num2)
logging.debug("Add : {} - {} = {}".format(num1, num2, add_result))

sub_result = sub(num1, num2)
logging.debug("Sub : {} - {} = {}".format(num1, num2, sub_result))


multi_result = multi(num1, num2)
logging.debug("Multipication : {} - {} = {}".format(num1, num2, multi_result))


division_result = devision(num1, num2)
logging.debug("Divison : {} - {} = {}".format(num1, num2, division_result))


#logging.debug() doesn't display anything by default.

#only sends the message to the logging system. By default, Python's logging level is usually WARNING, 
#so DEBUG messages are ignored.
```
