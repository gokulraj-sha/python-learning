Mini Project

1 Receive an email from the user

2 Validate the email

3 If it is invalid, Log an error in a file.

4 If it is valid, Clean and structure the email.

5 Log each step of the program

```python
from typing_extensions import Writer


def write_log(message):
    with open("/Users/gokulraj/Documents/VS/Python/10_Rough_note/log.txt","a") as file:
        file.write(message+ "\n")



def validate(email):
    return "@" in email and "." in email


def clean_data(email):
    email = email.lower().strip()
    username, domain = email.split("@")

    return {

        "Username":username,
        "Domain": domain
    }


write_log("Log Started")

email = input("Enter the email")


if not validate(email):
    write_log("Invalid email")

else:
    cleaned = clean_data(email)

    write_log(f"Data Processing{cleaned}")


write_log("Log Ended")



```

```python
class pirate:
    def __init__(self, name, bounty):
        self.name = name
        self.__bounty = bounty  # Encapsulation: name-mangled private attribute

    def bounty_list(self):
        print(f"Pirate, {self.name}'s current bounty is: {self.__bounty}")

    # Getter: returns the current bounty
    def get_bounty(self):
        return self.__bounty

    # Setter: updates the bounty
    def set_bounty(self, bounty):
        self.__bounty = bounty


class Emperor(pirate):
    def __init__(self, name, bounty, territory):
        super().__init__(name, bounty)
        self.territory = territory 


shanks = Emperor("Shanks", 4_000_000_000, "Elbaf")
shanks.bounty_list()
print("Territory:", shanks.territory)

luffy = pirate("Luffy", 1_500_000_000)

# Display the current bounty
luffy.bounty_list()

# Get the current bounty
print("Current bounty:", luffy.get_bounty())

# Update the bounty
luffy.set_bounty(2_000_000_000)

# Check the updated bounty
luffy.bounty_list()
print("Updated bounty:", luffy.get_bounty())
```
