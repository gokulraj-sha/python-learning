```python
country = input("Enter your country: ").lower()

match country:
    case "india":
        print("IN")

    case "america":
        print("USA")

    case "russia":
        print("RU")

    case _:
        print("Unknown")
```

```python

country =input("Enter the country :").lower()
match country:
    case 'india' | 'indian':  # we can also use logical operator
      print("IN")

    case 'united States':
      print("US")

    case 'egypt':
      print("EG")

    case 'germany':
      print("DE")

    case _:        #Default (else)
      print("Unknown")
```

```python
n = 5

for i in range(1, n+1):
    spaces = n - i  # for space
    stars = 2 * i - 1 # formula for *

    print(" " * spaces + "*" * stars)
```
