```python
class user:
  def __init__(self, username, email, password):
    self.username = username
    self._email = email #protected
    self.password = password

  @property  #"@property", standard eay of creating as getter method, same as "get_" method
  def email(self):
    print("Email accessed")
    return self._email

  @email.setter # (syntax: @name of that we want setter. setters)ame as "set_ method"
  def new_email(self, new_email):
    if "@" and "gmail.com" in new_email:
      self._email = new_email



user1 = user("goku", '  goku@gmail.com ', '000')
user1.new_email = "vegeta@gmail.com" #@ __.setter triggers whenever a new value is assigned
print(user1.email)

```

**2. Encapsulation**




**simple definition:**
* Encapsulation is an OOP concept that bundles data and the methods that operate on it inside a class, while restricting direct access to the internal data and providing controlled access through methods.

**Interview answer:**
* Encapsulation is the process of wrapping data and its related methods into a single unit, usually a class, and controlling access to the internal data to prevent unintended modification.

```python
class BankAccount:

    def __init__(self):
        self._balance = 0.0  # Internal data; access is controlled by the class

    @property
    def balance(self):
        return self._balance  # Provides controlled access to the balance

    def deposit(self, amount):
        # Controls how the balance can be increased
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self._balance += amount

    def withdraw(self, amount):
        # Controls how the balance can be decreased
        if amount <= 0:
            raise ValueError("Withdraw amount must be positive.")
        if amount >= self._balance:
            raise ValueError("Insufficient funds.")
        self._balance -= amount


account = BankAccount()  # Create a BankAccount object

print(account.balance)   # Access the balance through the class interface

account.deposit(1.99)    # Modify balance through a controlled method
print(account.balance)   # Read the balance through the class interface

account.withdraw(1)      # Modify balance through a controlled method
print(account.balance)   # Read the balance through the class interface

account.withdraw(100)   # Error: class prevents an invalid withdrawal


# Encapsulation = keep data inside the class + control how it is accessed/modified
# Encapsulation = bundling data and methods in a class while "controlling acces"s to the data
```

**A good real-world example of encapsulation:**

Bank account: The account's balance is internal data, and we control how it is modified through methods such as deposit and withdraw. This prevents invalid or unintended changes to the data.”

Remember:
Encapsulation → Protect/control the data

**3. Abstraction**
* Abstraction means hiding the internal implementation details and showing only the essential features to the user.

* Abstraction is an OOP concept that hides unnecessary implementation details and exposes only the essential functionality to the user.

* for Example :

```python
class Emailservice:

    # Internal methods: hide implementation details from the user
    def _connect(self):
        print("Connecting to email server")

    def _authentication(self):
        print("Authenticating")

    # Public method: exposes only the required functionality
    def send_Email(self):
        self._connect()
        self._authentication()
        print("Sending email...")
        self._disconnect()

    def _disconnect(self):
        print("Disconnected from email server")


email = Emailservice()

# User only needs to call this method; internal steps are hidden
email.send_Email()


# Abstraction = hide implementation details and expose only essential functionality
```

```python
#Here, the user only calls:

email.send_Email()

"""
They don't need to know that internally it performs:

          _connect()
              ↓
          _authentication()
              ↓
          send email
              ↓
          _disconnect()

"""
```

**A real-world example of abstraction**

Email service: The user only calls send_email() without needing to know the internal steps such as connecting to the server, authentication, and sending the request. The implementation details are hidden, and only the required functionality is exposed.”

4. Inheritance

* Inheritance is an OOP concept where a child class inherits the attributes and methods of a parent class, allowing code reuse and extension.

* Inheritance allows a child class to reuse and extend the attributes and methods of an existing parent class. It promotes code reusability and helps represent relationships between related classes.

```python
class vehical:

  def __init__(self, brand, model, year):
    self.brand = brand
    self.model = model
    self.year = year

  def start(self):
    print("Vehicle is Starting")

  def stop(self):
    print("Vehicle is Stoping")


# Child class inherits from Vehicle
class car(vehical):

  def __init__(self, brand, model, year, number_of_doors, number_of_wheels):
    super().__init__(brand, model, year)  # Reuse parent's __init__()
    self.number_of_doors = number_of_doors
    self.number_of_wheels = number_of_wheels


# Child class inherits from Vehicle
class bike(vehical):

  def __init__(self, brand, model, year, number_of_wheels):
    super().__init__(brand, model, year)  # Reuse parent's __init__()
    self.number_of_wheels = number_of_wheels


car1 = car("Ford", 'Focus ', 2008, 5, 4)
bike1 = bike("Honda", 'Scoopy ', 2008, 2)

print(car1.__dict__)
car1.start()  # Inherited method from Vehicle

print(bike1.__dict__)
bike1.start()  # Inherited method from Vehicle


# Inheritance = child class reuses and extends the attributes/methods of a parent class
# Inheritance = Reuse + Extend
```

**5. Polymorphism**

* Polymorphism allows the same method name to behave differently for different objects.


```python
class vehical:

  def __init__(self, brand, model, year):
    self.brand = brand
    self.model = model
    self.year = year

  def start(self):
    print("Vehicle is Starting")

  def stop(self):
    print("Vehicle is Stoping")
#---------------------------------------------------#

class Car(vehical):

  def __init__(self, brand, model, year, number_of_doors):
    super().__init__( brand, model, year)

  def start(self): #overriding for the main class, for that we to give the some method name
    print("Car is Starting")

  def stop(self):#overriding to be specific
    print("Car is Stoping")



#---------------------------------------------------#

class Motorcycle(vehical):

  def __init__(self, brand, model, year, number_of_wheels):
    super().__init__( brand, model, year)
  def start(self):#overriding
    print("Motorcycle is Starting")

  def stop(self):#overriding
    print("Motorcycle is Stoping")


#---------------------------------------------------#

vehicles = [
    Car("Ford", 'Focus ', 2008, 5),
    Motorcycle("Honda", 'Scoopy ', 2008, 2)
]

for vehicle in vehicles:
  if isinstance(vehicle, vehical):
    print(f"Inspecting {vehicle.brand} {vehicle.model}({type(vehicle).__name__})")
    vehicle.start()
    vehicle.stop()

  else:
    raise Exception("Object is not a valid vehicle")

```

#Regular Expression

Regex pattern → defines the text pattern we want to find

**finditer()** -> returns matching objects (Gives match objects)

Syntax:- re.finditer("stirng_to_find", string_varible)


```python
import re

log = """
User 10234 accessed file.
User 20456 downloaded data.
User 30987 logged out.
"""

matches = re.finditer(r"\d+", log)

for match in matches:
    print(match.group(), match.span())

# finditer() → returns Match objects for all matches
```

**Meta Characters**


| Symbol  | Meaning                   | Example              |
| ------- | ------------------------- | -------------------- |
| `.`     | Any single character      | `a.c` → `abc`, `a1c` |
| `^`     | Beginning of string       | `^ERROR`             |
| `$`     | End of string             | `\.csv$`             |
| `*`     | 0 or more                 | `ab*c`               |
| `+`     | 1 or more                 | `ab+c`               |
| `?`     | 0 or 1 / optional         | `colou?r`            |
| `{n}`   | Exactly n times           | `\d{3}`              |
| `{n,m}` | n to m times              | `\d{2,4}`            |
| `[ ]`   | Character set             | `[a-z]`              |
| `\`     | Escape / special sequence | `\d`, `\.`           |
| `\|`    | OR                        | `ERROR\|WARNING`     |
| `( )`   | Group                     | `(ERROR\|WARNING)`   |


```python
import re

text = "Error: User123 failed login. Error: User45 failed login."

# .  → matches any single character
print(re.findall(r"User.", text))
# The . represents one character,
#if we want more character along with searching character, we can add more "..."

```

```python
# ^  → pattern must start at the beginning

re.search(r"^ERROR", "ERROR: Database failed")
```

```python
# $  → pattern must end at the end
print(re.findall(r"login\.$", text))
```

```python
# *  → previous character: 0 or more times
print(re.findall(r"ab*c", "ac abc abbc abbbc"))
```

```python
# +  → previous character: 1 or more times
print(re.findall(r"ab+c", "ac abc abbc abbbc"))
```

```python
# ?  → previous character: 0 or 1 time (optional)
print(re.findall(r"colou?r", "color colour"))
```

```python
# { } → exact number / range of repetitions
print(re.findall(r"\d{3}", "12 123 456 7890"))
```

```python
# [ ] → set: match any ONE character from the set
print(re.findall(r"[aeiou]", "Data Engineering"))
```

```python
# [ ] → set: match any ONE character from the set
print(re.findall(r"[a-zA-Z0-9]", "Data Engineering @123"))
```

```python
# \  → escape special characters / create special sequences
print(re.findall(r"\.csv", "file.csv file.json"))
```

```python
# |  → OR: match either pattern
print(re.findall(r"ERROR|WARNING", "ERROR found, INFO found, WARNING found"))
```

```python
# ( ) → group related patterns together
print(re.findall(r"(ERROR|WARNING):", "ERROR: failed WARNING: retry"))

#Here, (ERROR|WARNING) groups the alternatives together, and : comes after the group.
#Grouping lets you treat multiple things as one unit.
```

**More Special Charaters:**



* \d : Matches any decimal digit; [0-9].

* \D : Matches any non-digit character;

* \s : Matches any whitespace character; (space " " tab "\t" newline "\n")

* \S : Matches any non-whitespace character;

* \w : Matches any alphanumeric (word) character; [a-zA-Z0-9_].

* \W : Matches any non-alphanumeric character;

* \b : Matches where the specified characters are at the beginning or at the end of a word;

* \B : Matches where the specified characters are present, but NOT at the beginning or end of a word.

```python
import re


# \d → matches digits
import re

log = "User ID: 10234 completed transaction 56789"

ids = re.findall(r"\d+", log)

print(ids)
```

```python
# \D → matches non-digit characters

phone = "Phone: +91-98765-43210"

numbers = re.findall(r"\D+", phone)

print(numbers)
```

```python
# \s → matches whitespace (space, tab, newline)
print(re.findall(r"\s+", text))

text = "Data    Engineering   Pipeline"
cleaned = re.sub(r"\s+", " ", text)

print(cleaned)
```

```python
# \S → matches non-whitespace characters
log = "User ID: 10234 completed transaction 56789"

print(re.findall(r"\S+", log))
```

```python
# \w → matches word characters (letters, digits, underscore)
text = "Users: data_engineer123, python_dev99"

users = re.findall(r"\w+", text)

print(users)
```

```python
# \W → matches non-word characters
text = "data-engineer/python Sql"

separators = re.findall(r"\W+", text)

print(separators)
```

```python
# \b → matches a word boundary
text = "ERROR ERROR_CODE NOERROR"
re.findall(r"ERROR", text)

#But maybe you only want the standalone word ERROR, not ERROR_CODE or NOERROR.
```

```python
# \b → matches a word boundary,
#searches for the match at the begin of a block(string or word )("E"rror)
log = "ERROR: database failed. ERROR_CODE: 500"

errors = re.findall(r"\ERROR\b", log)

print(errors)
```

```python
# \B → matches a position that is NOT a word boundary,
#searches for the match at anywhere at a  block(string or word )("E"rror)
texte = "user in the User_123"
print(re.findall(r"\Bser", "User_123"))
```

| Pattern | Meaning             | Example              |

| `\s`    | Whitespace          | space, `\t`, `\n`    |
| `\S`    | Non-whitespace      | `A`, `1`, `!`        |
| `\w`    | Word character      | letters, digits, `_` |

| `\B`    | Not a word boundary | `\BERROR`            |
