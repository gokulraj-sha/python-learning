# Merging_and_Joining Datasets

Different Types of joins
* Inner Join :- Gives Matching keys in both Tables 
* Left Join :- All Data from left and only matching from Right 
* Right Join :- All Data from Right and only matching from Left
* Outer Join :- Gives all the Data from both the table and marks unmatched rows as NaN

# Best Practices for Joins

* Specify the Parent Table First
* validate Keys
* Inspect Missing Data

***Use Case:-* Find all sales_rep who has made the sales and include only matching records**

```python
import pandas as pd
```

```python
# Sales Representative Table (Parent)
sales_reps = pd.DataFrame({
    "rep_id": [1, 2, 3],
    "name": ["Goku", "Luffy", "Guts"],
    "region": ["North", "South", "West"]
})
```

```python
# Sales Data Table (Child)

sales = pd.DataFrame({
    "sale_id": [101, 102, 103, 104 ],
    "sale_rep_id": [1, 2, None, 4], 
    "sale_amount": [500, 1000, 750, 1200]
})
```

```python
sales_reps
```

```python
sales
```

**`Inner Joins`**

```python
inner_join = pd.merge(
    sales_reps, #table_1
    sales,  #table_2
    left_on="rep_id", #primary_key of first table
    right_on="sale_rep_id", #forign_key of second table
    how="inner" #Joining method
)
```

```python
inner_join
```

**`Left Joins`**


**Use Case:-** Display the sales_reprenataive who has not made any sales

```python
left_join = pd.merge(
    sales_reps,
    sales,
    left_on= "rep_id",
    right_on = "sale_rep_id",
    how= "left"
)
```

```python
left_join
```

```python
#filtering the null rows using pandas .isnull()
inactive_reps = left_join[left_join["sale_id"].isnull()]
```

```python
inactive_reps
```

```python
#filtering the null rows using pandas sql-like query
left_join.query("sale_id.isnull()")
```

`Right Join`

**Use Case:-**  Find orphan sales records(child table has additional records with no value in parent table )

```python
right_join = pd.merge(
    sales_reps,
    sales,
    left_on="rep_id",
    right_on="sale_rep_id",
    how="right"
)
```

```python
right_join
```

```python
orphan_data = right_join[right_join["rep_id"].isnull()]
```

```python
orphan_data
```

```python
right_join.fillna({"name":"unknown", "region":"unknown", "sale_rep_id":"unknown"}, inplace=True)

#inplace=True → modify the original DataFrame directly
#inplace=False (default) → return a modified copy; original stays unchanged
```

`LEFT JOIN` → Keep all main data; add matching information from the other table

* Example: Keep all sales, add customer details. Missing customer → NaN

`RIGHT JOIN` → Keep all data from the other/master table; add matching information

* Example: Keep all employees, even those with no sales.


**`Outer Join`**

A full outer join in pandas keeps all records from both DataFrames, matching the common records and filling unmatched values with NaN.

```python
import pandas as pd
```

```python
# Sales Representatives Table (Parent)
sales_reps = pd.DataFrame({
    "rep_id": [1, 2, 3],
    "name": ["Alice", "Bob", "Charlie"],
    "region": ["North", "South", "West"]
})


# Sales Data Table (Child)
sales = pd.DataFrame({
    "sale_id": [101, 102, 103, 104],
    "sale_rep_id": [1, 2, None, 4],  # Includes a missing and an unmatched ID
    "sale_amount": [500, 1000, 750, 1200]
})
```

```python
#function
def calculate_range(series):
    return series.max() - series.min()

custom_aggegation = data.groupby("sale_rep_id")["sale_amount"]. \
    agg(
        total_sale = "sum",
        sales_min = "min",
        sales_max= "max",
        sales_range =calculate_range
    )
```

# Chaining  Transformation

Chaining transformations means performing multiple DataFrame operations one after another in a single chain, instead of creating many intermediate variables.

* Efficiency → fewer unnecessary steps/variables
* Readability → transformations are easy to follow
* Scalability → easier to add or modify transformations

**Use Case:- Calculate Commission Earned for Completed Sales**

```python
import pandas as pd


toyota_sales_data= pd.read_csv("Data/car_sales/toyota_sales_data.csv")
sales_reps_data= pd.read_csv("Data/car_sales/sales_reps_data.csv")
```

```python
# Normal_1 method
completed_sales = toyota_sales_data[toyota_sales_data["sale_status"] == "Completed"]
```

```python
# Normal_2 method
completed_sales["commission_earned"] = (
    completed_sales["sale_amount"] * completed_sales["commission_pct"]
)
```

`Or`

```python
#Using Chaining  Transformation
sales_with_commission = (
    toyota_sales_data[toyota_sales_data["sale_status"] == "Completed"]
    .assign(
        commission_earned=lambda x:
            x["sale_amount"] * x["commission_pct"]
    )
)
```

```python
sales_with_commission
```

`OR`

**Chaining Transformation using Multiple Datasets**

```python
sales_with_commission = (
    pd.merge(
        sales_reps_data,
        toyota_sales_data,
        
        left_on= "rep_id",
        right_on="sale_rep_id",
        how= "inner"
    )

    .query("sale_status == 'Completed'") \
    .assign(commision_earned = lambda x : x["sale_amount"] * x["commission_pct"])
)
```

```python
sales_with_commission
```

# Advance Chaining Transformation

**Real World Example**

```python
import pandas as pd


toyota_sales_data= pd.read_csv("Data/car_sales/toyota_sales_data.csv")
sales_reps_data= pd.read_csv("Data/car_sales/sales_reps_data.csv")
```

```python
sales_with_commission = (
    toyota_sales_data.fillna({"commission_pct" : 0})
    .query("sale_status == 'Completed'")
    .assign(
        commision_earned = lambda x: 
        round(x["sale_amount"] * x["commission_pct"], 2)
        )
)
```

```python
sales_with_commission
```

```python
#Using multiple Dataset
pd.merge(
        sales_reps_data,
        toyota_sales_data,
        left_on= "rep_id",
        right_on="sale_rep_id",
        how= "inner"
    ).fillna({"commission_pct" : 0}) \
    .query("sale_status == 'Completed'") \
    .assign(
        commision_earned = lambda x: 
        round(x["sale_amount"] * x["commission_pct"], 2)
        )
```

```python
#using Group by
#Using multiple Dataset
result= (
    pd.merge(
        sales_reps_data,
        toyota_sales_data,
        left_on= "rep_id",
        right_on="sale_rep_id",
        how= "inner"
    )
    .fillna({"commission_pct" : 0})
    .query("sale_status == 'Completed'")
    .assign(
        commission_earned = lambda x: 
            round(x["sale_amount"] * x["commission_pct"], 2)
    )
    .groupby("region")
    .agg(
        total_sales = ("sale_amount", "sum"),
        avg_commission = ("commission_earned","mean")
    )
    .reset_index()
)
```

```python
result
```

**Best practice: use () for multi-line chaining**

result = (

    pd.merge(...)

    .fillna(...)

    .query(...)

    .assign(...)

    .groupby(...)

    .agg(...)

    .reset_index()
    
)

# Complete Workflow

**Monthly Sales Commision**

```python
import pandas as pd

toyota_sales_data= pd.read_csv("Data/car_sales/toyota_sales_data.csv")
sales_reps_data= pd.read_csv("Data/car_sales/sales_reps_data.csv")
```

```python
monthly_summary = (
    pd.merge(
        sales_reps_data,
        toyota_sales_data,
        left_on= "rep_id",
        right_on="sale_rep_id", 
        how= "inner"
    )
    .fillna({"commission_pct": 0})
    .assign(total_commission = lambda x : round(x["sale_amount"] * x["commission_pct"]))
    .assign(sale_month = lambda x: pd.to_datetime(x["sale_date"]).dt.to_period("M")) #to display only months to calculate Total sales for each month
    .groupby(["sale_month", "rep_id", "first_name", "last_name", "email"])
    .agg(
        total_amount = ("sale_amount","sum"),
        total_commision = ("total_commission", "sum")
    )
    .reset_index()


)
```

```python
monthly_summary
```

```python
#write to CSV
monthly_summary.to_csv("data/car_sales/monthly_sales_commision.csv", index=False) # "index=False" to remove the index
```
