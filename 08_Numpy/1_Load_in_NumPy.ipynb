# NumPy

**NumPy (Numerical Python)** is a Python library used for **fast numerical calculations and working with arrays**.

* NumPy = **Arrays + Fast numerical calculations**
* Arrays can be:
  * 1D → `[1, 2, 3]`
  * 2D → rows and columns
  * 3D → collection of 2D arrays

---

## Why NumPy instead of Python Lists?

NumPy is preferred for numerical data because its arrays are **more memory-efficient and faster for numerical operations**.

| Python List | NumPy Array |
|---|---|
| General-purpose | Designed for numerical calculations |
| Can store different data types | Usually stores one fixed data type |
| More memory usage | More memory efficient |
| Slower for large numerical calculations | Faster for numerical calculations |
| Often requires loops | Supports direct array operations |

**Key idea:**  
NumPy can perform operations on an **entire array at once**, instead of manually writing a loop.

---

## Why is NumPy Faster?

### 1. Fixed Type

* NumPy arrays use a **fixed data type** such as `int32`.
* Each element has a fixed size in memory.
* No repeated type checking is needed while processing elements.
* This reduces memory overhead and improves speed.

Python lists store additional information for each object, such as:
* Size
* Reference count
* Object type
* Object value

**Memory:**  
`Fixed type → Less overhead → Faster processing`

### 2. Contiguous Memory

* NumPy stores array elements **next to each other in memory**.
* This makes data access faster.
* It allows efficient **CPU cache utilization**.
* It also enables **SIMD vector processing** → processing multiple values at once.

**Memory:**  
`Contiguous memory → Faster data access → Faster calculations`

* `a.itemsize` → bytes used by **one element**
* `a.nbytes` → bytes used by **the entire array**

**Formula:**
`nbytes = itemsize × number of elements`

````python
import numpy as np

arr = np.zeros((5, 5), dtype=int)

arr[0, :] = 1    # top row
arr[-1, :] = 1   # bottom row
arr[:, 0] = 1    # first column
arr[:, -1] = 1   # last column

arr[2, 2] = 9    # center

print(arr)
````

````python
#Same Practice
arr = np.zeros((5,5), dtype=int)

arr[0,:] = 1
arr[:,0] = 1
arr[-1,:] = 1
arr[:, -1] =1

arr[2,2] = 2

print(arr)
````

#### **Be Careful when copying arrays!!!**

````python
a = np.array([1,2,3])
b = a
b[0] = 100

print(a)
print(b)
#It's just assigns, so if we modify the b, the actual main data a also get chnages, so we use "copy()"
````
