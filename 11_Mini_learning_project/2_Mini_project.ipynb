### **Requirements:**
1. Calculate the total revenue for each record using price * quantity.
2. Calculate the overall revenue across all records.
3. Count how many records have a quantity greater than zero.
4. Identify records where the quantity is zero.
5. Write a function called analyze_sales() that performs these calculations and returns the results.

```python
sales = [
    {"product": "Laptop", "price": 50000, "quantity": 2},
    {"product": "Mouse", "price": 500, "quantity": 5},
    {"product": "Keyboard", "price": 1500, "quantity": 3},
    {"product": "Laptop", "price": 50000, "quantity": 1},
    {"product": "Mouse", "price": 500, "quantity": 0},
]
```

```python
def analyze_sales(sales):

    overall_revenue = 0
    total_records = 0
    zero_records = 0

    for items in sales:
        print(f"Total Revenu for {items['produc']} is {items['price']* items['quantity']}")

    for items in sales:
        overall_revenue += items["price"] *  items["quantity"]

    print("Overall revenue: ", overall_revenue)

    for items in sales:
        if items["quantity"]>0:
            total_records += 1 
    print("Records with quantity greater than zero: ",total_records)

    for items in sales:
        if items["quantity"]==0:
            zero_records += 1
    print("Records with zero quantity: ", zero_records) 

    return overall_revenue, total_records, zero_records

analyze_sales(sales)
```

### **Requirements:**
Write a function called clean_employees(employees) that:
1. Identifies employees whose age is missing (None).
2. Identifies employees whose salary is missing (None).
3. Counts how many employees have both age and salary available.
4. Returns these three results.


```python
employees = [
    {"id": 1, "name": "Rahul", "age": 25, "salary": 30000},
    {"id": 2, "name": "Priya", "age": None, "salary": 35000},
    {"id": 3, "name": "Arun", "age": 28, "salary": None},
    {"id": 4, "name": "Sneha", "age": 24, "salary": 32000},
    {"id": 5, "name": "Kiran", "age": None, "salary": None},
]
```

```python
def clean_employees(employees):
    
    missing_age=0
    missing_salary=0
    both=0

    for emp in employees:
        if emp['age'] is None:
            missing_age += 1

        if emp['salary'] is None:
            missing_salary += 1
            
        if  (emp['age'] is  not None ) and (emp['salary'] is not None):
            both += 1
            

    print("Employees with missing age", missing_age)
    print("Employees with missing salary", missing_salary)
    print("Employees with both values available", both)

    return missing_age, missing_salary, both


clean_employees(employees)

    

    

    
```
