```python
def category(sale):
    if sale["revenue"] < 8000:
        sale["category"] = "Low"

    elif sale["revenue"] >= 8000 and sale["revenue"] <= 15000:
        sale["category"] = "Medium"

    elif sale["revenue"] > 15000:
        sale["category"] = "High"

    return sale["category"]
```

Functions to revise:

Important: Check the assumptions behind the merge. A region name must match between the two DataFrames, and inconsistent capitalization can prevent a match.

```python
reprensentatives = pd.DataFrame({
    "region": ["North","South","East","West"],
    "sales_rep":["Arun","Priya","Rahul","Meena"],
    "target":[120000, 80000, 60000, 100000]
})
```

```python
reprensentatives["region"] = reprensentatives["region"].str.upper()
```

```python
reprensentatives
```

```python
#1
merged_data = pd.merge(
    sales,
    reprensentatives,
    left_on="region",
    right_on="region",
    how = "outer"
)
merged_data
```

```python
#2 Unmtched
unmatched = pd.merge(
    sales,
    reprensentatives,
    on="region",
    how="left",
    indicator=True
)

unmatched = unmatched[unmatched["_merge"] == "left_only"]
unmatched
```

```python
#3
rev_sales_rep = (
    merged_data
    .groupby("sales_rep")["revenue"]
    .sum()
    .reset_index()
)
rev_sales_rep
```

```python
#4
# Step 1: Calculate total revenue for each region
com_regional_vs_target = (
    merged_data
    .groupby(["region", "target"])["revenue"]
    .sum()
    .reset_index()
)

# Step 2: Calculate the difference between revenue and target
com_regional_vs_target["difference"] = (
    com_regional_vs_target["revenue"] - com_regional_vs_target["target"]
)

# Step 3: Display the result
com_regional_vs_target
```

```python
summary_table_c = (
    sales
    .groupby("region")
    .agg(
        total_revenue = ("revenue", "sum"),
        average_sale = ("revenue","mean")
    )
    .reset_index()
)
summary_table_c
```
