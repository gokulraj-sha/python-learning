**Use Case :If the tottal_commison is greater than 100, return True**

```python
import pandas as pd
toyota_data = pd.read_csv(
    "Data/car_sales/toyota_sales_data.csv"
    )
```

```python
#toyota_data.fillna({"commission_pct":0}, inplace=True)
```

```python
#calculate total_commision and remove null

def total_commision_calulation(sale):
    commission_pct = sale["commission_pct"] if pd.notnull(sale["commission_pct"]) else 0
    return sale["sale_amount"] * commission_pct
```

```python
toyota_data.apply(total_commision_calulation, axis=1)
```

`Or`

```python
toyota_data["commisssion_total"] = toyota_data.apply(total_commision_calulation, axis=1)
```

```python
def high(sale):
    return sale["commisssion_total"]>1000
```

```python
toyota_data.apply(high, axis=1)
```

```python
toyota_data["greates_than_1000"] = toyota_data.apply(high, axis=1)
```
