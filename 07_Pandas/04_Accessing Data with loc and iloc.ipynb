## ".loc" (Lable-Based Indexing)

.loc is a label-based indexer in pandas. It allows us to select or modify specific rows and columns using their labels, and it's also commonly used for conditional filtering.



```python
import pandas as pd

# loading the data
toyota_data = pd.read_csv("/Users/gokulraj/Documents/VS/Project/autosales_insights/Data/car_sales/toyota_sales_data.csv")
sales_rep_data = pd.read_csv("/Users/gokulraj/Documents/VS/Project/autosales_insights/Data/car_sales/sales_reps_data.csv")

```

```python
#To get the rows based on the requirements usind index

toyota_selection = toyota_data.loc[5:9] #index 5 to 9
toyota_selection
```

```python
# if we want perticular columns with required rows

toyota_selection = toyota_data.loc[5:9, ['sale_date', 'car_model', 'sale_amount']] 
toyota_selection
```

---

To get the amonunt greater than 30K

```python
high_sales = toyota_data.loc[toyota_data['sale_amount'] > 30000]
high_sales
```

---

```python
#we can also do this without "loc"
toyota_data.loc[toyota_data['sale_amount'] > 30000]

#But we can't control over both rows and columns.
#Like: ['car_model', 'sale_amount', 'sale_date'] in the bolow example
```

To get the amonunt greater than 30K along with car_model, sale_amount, sale_date

```python

high_sales_model = toyota_data.loc[
    toyota_data['sale_amount'] > 30000, 
    ['car_model', 'sale_amount', 'sale_date']
]
high_sales_model
```

---

To get the amonunt greater than 30K and car_model is "RAV4"

```python
high_rava4_sales = toyota_data.loc[
    (toyota_data['sale_amount']> 3000) & (toyota_data["car_model"] == "RAV4"),
    # ()&()
    ['car_model', 'sale_amount', 'sale_date']
    ]

high_rava4_sales
```

---

Convert all the "Null" values to "0"

```python
toyota_data.loc[
    toyota_data["commission_pct"].isnull(),
    "commission_pct"
] = 0.0
```

```python
toyota_data
```
