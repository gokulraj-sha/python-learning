## **pd.concat()**

**pd.concat() combines two or more pandas Series or DataFrames by stacking them vertically or placing them horizontally.**

**1. Vertical concatenation — axis=0**

The rows are stacked one below another. This is the default behavior.

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Arun", "Bala"],
    "Salary": [25000, 30000]
})

df2 = pd.DataFrame({
    "Name": ["Chitra", "Deepak"],
    "Salary": [28000, 32000]
})

result = pd.concat([df1,df2],ignore_index=True)

result
```

**2. Horizontal concatenation — axis=1**

The columns are placed side by side, aligning rows by their index.

```python
df1 = pd.DataFrame({
    "Name": ["Arun", "Bala"]
})

df2 = pd.DataFrame({
    "Salary": [25000, 30000]
})

result = pd.concat([df1, df2], axis=1)

print(result)
```

| Feature           | `concat()`                                   | `merge()`                                             |
| ----------------- | -------------------------------------------- | ----------------------------------------------------- |
| Purpose           | Combines tables by stacking or aligning them | Combines tables using matching key columns or indexes |
| Best for          | Appending batches of similar records         | Joining related datasets                              |
| Common parameters | `axis=0` or `axis=1`                         | `on="ID"`, `how="inner"`                              |
| Example use       | Combining January and February sales records | Combining sales data with salesperson details         |

## **JSON**

JSON (JavaScript Object Notation) is a text-based format for storing and exchanging structured data. It is commonly used in APIs and data pipelines.

```python
import pandas as pd

employees = [
    {"name": "Arun", "salary": 25000},
    {"name": "Bala", "salary": 30000},
    {"name": "Chitra", "salary": 28000}
]

pd.DataFrame(employees).to_json(
    "employees.json",
    orient="records",
    indent=4
)
```

```python
df = pd.read_json("employees.json")

print(df)
```

## **Parquet**
Parquet is a column-oriented binary file format designed for efficient data storage and processing.

Compared with CSV, Parquet often offers:
- Smaller file sizes through compression.
- Efficient reading of selected columns.
- Better integration with tools such as Apache Spark, AWS Athena, and data lakes.

```python

import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Bala", "Chitra"],
    "Salary": [25000, 30000, 28000]
})

df.to_parquet("test.parquet", index=False)

print("Parquet file created successfully!")

result = pd.read_parquet("test.parquet")
print(result)
```
