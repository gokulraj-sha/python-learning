# String Operation

```python
import pandas as pd

data = pd.read_csv("Data/car_sales/sales_reps_data.csv")
data.head()
```

```python
data.shape
```

```python
data["first_name"].str.strip() #Remove extra spaces
data["email"].str.lower()  #Convert to lowercase
data["region"].str.upper()  #Convert to uppercase
data["email"].str.contains("gmail", na=False)  #Find matching text
data["region"].str.replace("North", "N", regex=False) #Replace text
```
