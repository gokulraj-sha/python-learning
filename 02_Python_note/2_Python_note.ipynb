**4. Orchestrator Function**

Controls program flow by calling other functions in the correct order.

Mini Project

1 Receive an email from the user

2 Validate the email

3 If it is invalid, Log an error in a file.

4 If it is valid, Clean and structure the email.

5 Log each step of the program

```python
from google.colab import drive
drive.mount('/content/drive')

```

```python
from typing_extensions import Writer


# Action Function
def write_log(message):
    # Task: store application log messages in a file on drive C
    with open("/content/drive/MyDrive/Projects/mini_project.log", "a") as file:
      file.write(message + "\n")


# Validation Function
def is_valid_email(email):
    return "@" in email and "." in email


#Transformation Function
def clean_and_split_email(email):
    # Task: clean an email and split it into username and domain
    email = email.strip().lower()
    username, domain = email.split("@")
    return {
        "username": username,
        "domain": domain
    }



write_log("App Started")

#We receive an email from a user
email= input("Enter your Email:")

#We must check if it is valid

#If it is not valid, we log the problem
if not is_valid_email(email):
  write_log(f"Invalid Email recieved: {email}")
else:
  clean = clean_and_split_email(email) # creating a 'clean' variable because, we want to print or write it.
  write_log(f"Processed Email: {clean}")


write_log("App Ended")
#If it is valid, we clean it and store structured information
#And we log what happened
```

Using **Orchestrator Function**




```python
# Orchestrator Function
def process_user_email(email):
    write_log("App Started")

    #We must check if it is valid
    #If it is not valid, we log the problem
    if not is_valid_email(email):
        write_log(f"Invalid Email received: {email}")

    #If it is valid, we clean it and store structured information
    else:
        clean_email = clean_and_split_email(email)
        write_log(f"Processed Email: {clean_email}")

    write_log("App Stopped")





#We receive an email from a user
email = input("Please enter Your Email: ")
process_user_email(email)
```

**ISO 8601**

A standard format for representing date and time so different systems can understand it consistently.

**1. isoformat()**


Definition: Converts a Python datetime object into an ISO 8601 formatted string.

```python
import datetime as dt

now = dt.datetime.now()

print(now.isoformat())
```

**2. fromisoformat()**
Definition: Converts an ISO 8601 formatted string back into a Python datetime object.

```python
import datetime as dt

date_string = "2026-09-09T22:10:30"

date_time = dt.datetime.fromisoformat(date_string)

print(date_time)
print(type(date_time))
```

Why? : Because APIs, databases, logs and cloud systems commonly exchange timestamps in standardized formats.

isoformat() → datetime → string
fromisoformat() → string → datetime

**Unix / Epoch Timestamp**

A number representing the number of seconds since January 1, 1970 UTC.

**1. timestamp()**

Definition: Converts a datetime object into a Unix/Epoch timestamp, which is the number of seconds since January 1, 1970 UTC.

```python
import datetime as dt

now = dt.datetime.now()
timestamp = now.timestamp()

print(now)
print(timestamp)
```

Why is this useful?

Computers can easily compare numeric timestamps.

EX:

```python
time1 = 1788963030
time2 = 1788966630

print(time2 - time1)
```

**2. fromtimestamp():** does the reverse:

number → datetime

```python
import datetime as dt

timestamp = 1788963030

date_time = dt.datetime.fromtimestamp(timestamp)

print(date_time)
```

datetime → timestamp() → Unix number

Unix number → fromtimestamp() → datetime

**Generator**

* A generator is a special type of iterator (Special way to create an intirator) that produces values one at a time using yield.
*  In iterator we use iter()and next() funtion bu ourself, but by using Generator, we can make pyhton do it for us
* **"Yield"** is a special keyword that make our function as generator
* It uses lazy evaluation, so values are generated only when needed, making generators memory-efficient for processing large amounts of data.

return → gives a result and ends the function.

yield → gives a result and pauses the function.

**`Generator vs List `**


```python
#Consider this:
numbers = [1, 2, 3, 4, 5]
#The **list stores** all the values.



#A generator:
def numbers():
    for i in range(1, 6):
        yield i
#produces them one at a time.
```

**Generator advantages**
1. ✅ Memory efficient
Doesn't need to hold the entire generated sequence in memory.
2. ✅ Lazy
Produces values only when needed.
3. ✅ Good for large data

Useful for:
* large files
* logs
* ETL processing
* data streams
* large datasets


Imagine a file containing: 10 GB

You don't necessarily want to load the entire file into memory at once.

Instead of:

data = load_entire_file()



```python
#you can process data one line at a time:
def read_data(file):
    for line in file:
        yield line
```

```python
#Then:
for line in read_data(file):
    process(line)
```

This is called lazy evaluation.

```python
#Example:

def load_images():
    for i in range(1, 6):
        print(f"Loading image {i}")
        yield i


images = load_images()

print(next(images))  # Loading image 1 → 1
print(next(images))  # Loading image 2 → 2
print(next(images))  # Loading image 3 → 3

# We stop here...

print("Continue:")
print(next(images))  # Loading image 4 → 4
print(next(images))  # Loading image 5 → 5
```

```python
def sqare():
    n = 1
    while n< 10:
      eq = n*n
      yield eq
      n +=1

sq = sqare()
for i in sq:
  print(i)
```

```python
#Real World Example:
def read_logs(file_path):
    with open(file_path, "r") as file:
        for line in file:
            yield line


logs = read_logs("server_logs.txt")

# Process first 3,000 log records
for _ in range(3000):
    log = next(logs)
    print(log)

# Suppose the pipeline stops here...
# Later, continue from the next record

log = next(logs)
print(log)






from IPython.display import Javascript
display(Javascript('''google.colab.output.setIframeHeight(0, true, {maxHeight: 300})'''))
```
