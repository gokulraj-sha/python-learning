**Shallow copy**



Shallow copy creates a new outer object, but nested objects are still shared. Deep copy creates a completely independent copy, including nested objects.

```python
list1 = [['a','b'], ['c','d']]
list2= list1.copy()

print("Original list:", list1)
print("Copied list:", list2)
```

```python
list1.pop()
print("Original list:", list1)
print("Copied list:", list2)
```

```python
list2[0].append("z")
print("Original list:", list1)
print("Copied list:", list2)
```

**Deep copy**

 Deep copy creates a completely independent copy, including nested objects.

```python
import copy
list1 = [['a','b'], ['c','d']]
list2= copy.deepcopy(list1)

print("Original list:", list1)
print("Copied list:", list2)
```

```python
list2[0].append("z")
print("Original list:", list1)
print("Copied list:", list2)
```

Compare the the original and copy

```python
original = [
    ['a', 'b'],      # Row 0
    ['c', 'd'],      # Row 1
]

#Assignment
copy1 = original
print("Same Object?", original is copy1, "\n")

#Shallow Copy
copy2 = original.copy()
print("Same Object?", original is copy2)
print("Shared Lists?", original[0] is copy2[0], "\n") # Nested object is not copied and refers to the original, so True

#Deep Copy
copy3 = copy.deepcopy(original)
print("Same Object?", original is copy3)
print("Shared Lists?", original[0] is copy3[0], "\n") #Also copies nested obeject, so False
```

Order Lists in Python **sort(), sorted(), reverse()**

```python
matrix = [
    ['h', 'd', 'a'],    # Row 0
    ['i', 'b', 'g'],    # Row 1
    ['c', 'f', 'e']     # Row 2
]
matrix.sort()"""sort() modifies the original list, that's the reason
#we can't assign it to a different variable, like "m1=matrix.sort()"""
print("Row sorted",matrix) #sorts the line by line , besed on 1st element(h, i, c), only rows

matrix.sort(reverse=True)
print("Row sorted in reverse",matrix)

#To sort in nested list
matrix[0].sort()
matrix[1].sort()
matrix[2].sort()
print("Sorted nested list",matrix)

#To sort in nested list in Reverse
matrix[0].sort(reverse= True)
matrix[1].sort(reverse= True)
matrix[2].sort(reverse= True)
print("Sorted nested list in reverse",matrix)

```

**sort()** modifies the original list, while **sorted()** returns a new sorted list.

```python
letters= ['c', 'a', 'b']

new_list = sorted(letters) '''while sorted() returns a new sorted
list, that the reason we are able to assign it a new variable'''
print("Original List: ", letters)
print("New Sorted List: ",new_list)
```
