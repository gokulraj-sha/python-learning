| Level | Meaning | Simple way to remember |
|---|---|---|
| `DEBUG` | Detailed information useful while developing/debugging | 🔍 **What is happening?** |
| `INFO` | Normal information confirming things are working | ✅ **Everything is OK** |
| `WARNING` | Something unexpected happened, but the program can continue | ⚠️ **Be careful** |
| `ERROR` | A problem occurred and some operation failed | ❌ **Something failed** |
| `CRITICAL` | A very serious problem that may stop the program | 🚨 **Program may stop** |


| Attribute | Format | Simple description |
|---|---|---|
| `args` | Fromat yourself | Arguments passed to the logging call |
| `asctime` | `%(asctime)s` | Time when the log was created |
| `created` | `%(created)f` | Time when the log was created (timestamp) |
| `exc_info` | Fromat yourself | Exception/error information |
| `filename` | `%(filename)s` | Name of the source file |
| `funcName` | `%(funcName)s` | Name of the function making the log call |
| `levelname` | `%(levelname)s` | Logging level: DEBUG, INFO, WARNING, ERROR, CRITICAL |
| `levelno` | `%(levelno)s` | Numeric value of the logging level |
| `lineno` | `%(lineno)d` | Line number where logging was called |
| `module` | `%(module)s` | Name of the Python module |
| `msecs` | `%(msecs)d` | Milliseconds when the log was created |
| `message` | `%(message)s` | Actual log message |
| `msg` | Fromat yourself | Original message passed to logging |
| `name` | `%(name)s` | Name of the logger |
| `pathname` | `%(pathname)s` | Full path of the source file |
| `process` | `%(process)d` | Process ID |
| `processName` | `%(processName)s` | Process name |
| `relativeCreated` | `%(relativeCreated)d` | Time since logging was started (ms) |
| `stack_info` | Fromat yourself | Stack information about where logging was called |
| `thread` | `%(thread)d` | Thread ID |
| `threadName` | `%(threadName)s` | Thread name |
