**absolute()**

To get absolute path including current file


```python

print(my_file.parent.absolute()) 
#or
# .parent → moves one level up to the containing directory
print(my_dir.absolute().parent)


# .absolute() → returns the complete absolute path, including the current file/folder
print(my_dir.absolute())
```

**resolve()**

absolute() and resolve() soes the same thing except

Ex:

```python


# absolute() → converts to absolute path but may keep ".." or "."
p = Path("..").absolute()

# resolve() → converts to absolute path and resolves "..", "." and symlinks
p1 = Path("..").resolve()
print(p)
print(p1)


# .. → parent directory (one level above the current directory)
# . → current directory
# .. → parent directory
```

**"__file__ "** 
* current Python script's path
* resolve() → converts it to the complete absolute path


```python
p1 = Path(__file__).resolve() 
#since we are in ipynb file, it shows error, it will show if we are working on .py file
print(p1)

#output:- example.py
```

Search folder and all subfolders for names containing "txt"

```python
from pathlib import Path

path_lib_module = Path.home() / "Documents" / "VS" / "Path_lib_module"  # Start from Home → go to Path_lib_module

for p in path_lib_module.rglob("*txt"):  # Recursively search this folder and all subfolders for names containing "txt"
    print(p)  # Print each matching file path


#for p in path_lib_module.rglob("*exmaple",case_sensitive = False ):
# "case_sensitive = False " gives all even if it is .json, sh, txt all kind of file matching "example" regardless of file type
```

Note:- 
* *txt *  → contains "txt" anywhere
* txt*   → starts with "txt"
* *txt   → ends with "txt"
